## Load the frozen pre-lock sample
No actual game results are included in this sample, so this cell cannot hindsight-fit to the completed game.

```python
import json, sys
sys.path.insert(0, str(Path('src').resolve()))
import showdown_ceiling_game_script_v1 as ceiling
import showdown_adaptive_portfolio_v1 as adaptive
data=json.loads(Path('data/den_kc_prelock_sample.json').read_text())
print(data['slate_id'], data['game'], len(data['players']))
```

## Inspect TournamentP90 / ceiling adjustments

```python
rows=ceiling.apply(data['players'])
for p in rows:
    print(f"{p['Player']:<22} base={p['BaseP90']:>6.2f} tournament={p['TournamentP90']:>6.2f} boost={p['CeilingBoost']:>5.2f}  {p['ScriptTags']}")
```

## Inspect pregame captain-strength signal

```python
cmap={p['Player']:p for p in rows}
players={}
for raw in data['players']:
    q=cmap[raw['Player']]
    players[raw['Player']]=adaptive.Player(name=raw['Player'],team=raw['Team'],projection=raw['Projection'],p90=q['TournamentP90'],tournament_fit=raw.get('TournamentFit',50),cpt_leverage=raw.get('GSCptLeverage',100),cpt_ownership=raw.get('GSCptOwnershipPct',0),role_tier=raw.get('RoleTier','UNKNOWN'),role_confidence=raw.get('RoleConfidence',0),fragility=raw.get('FragilityClass','MEDIUM'))
for name,p in sorted(players.items(),key=lambda kv:adaptive.captain_strength(kv[1]),reverse=True):
    print(f"{name:<22} {adaptive.captain_strength(p):.3f}")
```
